### 📊 Cluster-wise Winner Distribution Visualization

This notebook generates a visualization of **model preference distribution across semantic task clusters**. Each prompt in the dataset is associated with one or more topics, and the goal is to identify which language model tends to win within each cluster of prompts.

**Key steps:**

1. **Topic Embedding and Clustering**:
   - Unique topics from the dataset are encoded using the `all-MiniLM-L6-v2` SentenceTransformer model.
   - UMAP is applied to reduce embedding dimensionality.
   - K-Means (`k=30`) is used to form semantically coherent task clusters.
   - Prompts are assigned a cluster ID by majority voting across the three topic modeling fields.

2. **Winner Attribution**:
   - For each prompt, the winning model is identified based on human preference (`model_a`, `model_b`, or tie).
   - Votes are aggregated within each cluster to determine which models win most frequently.

3. **Visualization**:
   - A grid of bar charts is generated, where each subplot corresponds to a cluster and shows the distribution of model wins.
   - The figure is saved as `visual/all_clusters.png`.

This analysis helps us understand model performance at a **task-level granularity**, revealing how different LLMs perform across distinct conceptual domains.


```python
import pandas as pd
import numpy as np
from collections import Counter
from sentence_transformers import SentenceTransformer
from umap import UMAP
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import re
from sklearn.neighbors import NearestCentroid
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import calinski_harabasz_score, silhouette_score
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_data_for_tasks_deciphered.csv"
)
all_topics = pd.unique(df[['topic_modeling_1', 'topic_modeling_2', 'topic_modeling_3']].values.ravel())
all_topics = [t for t in all_topics if pd.notnull(t)]

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
umap_model = UMAP(n_neighbors=100, n_components=5, min_dist=0.25, metric='cosine', random_state=42)
scaler = StandardScaler()
kmeans = KMeans(
    n_clusters=30,
    init='k-means++',
    n_init=50,
    max_iter=1000,
    tol=1e-6,
    algorithm='elkan',
    random_state=42
)
embeddings = embedding_model.encode(all_topics, show_progress_bar=True)
reduced = umap_model.fit_transform(embeddings)
scaled = scaler.fit_transform(reduced)
labels = kmeans.fit_predict(scaled)
topic_to_cluster = dict(zip(all_topics, labels))
def map_unknown(topic):
    emb = embedding_model.encode([topic])
    emb_umap = umap_model.transform(emb)
    emb_scaled = scaler.transform(emb_umap)
    return int(kmeans.predict(emb_scaled)[0])

def map_topic(topic):
    if pd.isnull(topic):
        return np.nan
    return topic_to_cluster.get(topic, map_unknown(topic))

for i in (1, 2, 3):
    df[f"topic_modeling_{i}_cluster"] = df[f"topic_modeling_{i}"].apply(map_topic)

def vote_clusters(row):
    clusters = [row[f"topic_modeling_{i}_cluster"] for i in (1, 2, 3) if pd.notnull(row[f"topic_modeling_{i}_cluster"])]
    if not clusters:
        return np.nan
    cnt = Counter(clusters)
    top_count = cnt.most_common(1)[0][1]
    winners = [c for c, count in cnt.items() if count == top_count]
    return winners[0] if len(winners) == 1 else "tie"

df["cluster_id"] = df.apply(vote_clusters, axis=1)
```

```python
import os
import math
import matplotlib.pyplot as plt
import numpy as np

def sort_key(x):
    try:
        return (0, int(x))
    except:
        return (1, str(x))

cluster_ids = sorted(
    [cid for cid in df['cluster_id'].dropna().unique() if cid != 'tie'],
    key=sort_key
)
cols = 5
rows = math.ceil(len(cluster_ids) / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols * 4, rows * 4))
axes = axes.flatten()
for ax, cid in zip(axes, cluster_ids):
    group = df[df['cluster_id'] == cid]
    counts = {}
    for _, row in group.iterrows():
        w = row['winner']
        if w == 'model_a':
            m = row['model_a']; counts[m] = counts.get(m, 0) + 1
        elif w == 'model_b':
            m = row['model_b']; counts[m] = counts.get(m, 0) + 1
        else:  
            for prefix in ('model_a', 'model_b'):
                m = row[prefix]; counts[m] = counts.get(m, 0) + 1

    models = list(counts.keys())
    votes = [counts[m] for m in models]
    ax.bar(models, votes)
    ax.set_title(f'Cluster {cid}')
    ax.set_xticklabels(models, rotation=45, ha='right')

for ax in axes[len(cluster_ids):]:
    ax.axis('off')

plt.tight_layout()

os.makedirs('visual', exist_ok=True)
out_path = 'visual/all_clusters.png'
plt.savefig(out_path, dpi=300)
plt.close(fig)

print(f"Saved combined plot to {out_path}")
```
